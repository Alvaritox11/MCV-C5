```python
import json
import os
import matplotlib.pyplot as plt
from PIL import Image

# 1. Update this path to where your actual VizWiz validation images are stored!
IMAGE_DIR = "/ghome/group05/datasets/VizWiz/train" 

# 2. Path to the detailed metrics you just generated
METRICS_FILE = "/ghome/group05/gerard/MCV-C5/Task3/metrics_computed/metrics_baseline-resnet18-gru-char.json"

with open(METRICS_FILE, 'r') as f:
    results = json.load(f)

print(f"Loaded {len(results)} evaluated predictions.")
```

```python
def plot_example(item, title_prefix=""):
    img_id = item['image_id']
    
    # VizWiz images usually have a specific naming convention. 
    # Adjust the filename format if yours is different (e.g., adding '.jpg' directly if img_id includes it)
    # Example format: VizWiz_val_0000xxxx.jpg
    img_filename = f"{img_id}" if str(img_id).endswith('.jpg') else f"VizWiz_train_{str(img_id).zfill(8)}.jpg"
    img_path = os.path.join(IMAGE_DIR, img_filename)
    
    try:
        img = Image.open(img_path).convert("RGB")
    except FileNotFoundError:
        print(f"❌ Could not find image at: {img_path}")
        return

    # Extract metrics
    avg_score = item['Avg_Score']
    meteor = item['METEOR']
    bleu4 = item['BLEU-4'] # is bleu-2 actually
    rouge = item['ROUGE-L']
    
    pred = item['prediction']
    refs = "\n".join([f"- {r}" for r in item['references'][:3]]) # Show up to 3 references

    # Plotting
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.imshow(img)
    ax.axis('off')
    
    # Text formatting
    title_text = (
        f"{title_prefix} (ID: {img_id})\n"
        f"Avg Score: {avg_score:.2f} | METEOR: {meteor:.2f} | BLEU-2: {bleu4:.2f} | ROUGE-L: {rouge:.2f}\n\n"
        f"PREDICTION:\n{pred}\n\n"
        f"GROUND TRUTH:\n{refs}"
    )
    
    ax.set_title(title_text, fontsize=12, loc='left', pad=10, fontweight='bold')
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def generate_waterfall_improvement_plot():
    # =========================================================================
    # 1. ⬇️ PUT YOUR ACTUAL METEOR SCORES HERE ⬇️
    # Order: Baseline, +ResNet-50, +Word Representation, +Teacher Forcing, +Hyp Opt
    # =========================================================================
    METEOR_SCORES = [6.799, 7.940, 18.971, 33.988, 35.900]  
    
    # Define the labels for the x-axis
    labels = [
        "Baseline\n(ResNet-18 + GRU)", 
        "+ ResNet-50\nEncoder", 
        "+ Word Level\nTokens", 
        "+ Teacher\nForcing", 
        "+ Hyp\nOptimization"
    ]

    # Established pastel aesthetic
    colors = ["#FF7F6E", "#C7A4FF", "#6CB4EE", "#8ED6A8", "#FFB7B2"]

    fig, ax = plt.subplots(figsize=(11, 6))
    
    # 2. Calculate the heights (the delta) and the bottoms for the floating bars
    heights = [METEOR_SCORES[0]]
    bottoms = [0.0]
    
    for i in range(1, len(METEOR_SCORES)):
        heights.append(METEOR_SCORES[i] - METEOR_SCORES[i-1]) # Only plot the improvement
        bottoms.append(METEOR_SCORES[i-1])                    # Start where the last one ended
        
    x_positions = np.arange(len(labels))
    bar_width = 0.6
    
    # Plot the floating bars
    bars = ax.bar(x_positions, heights, bottom=bottoms, color=colors, edgecolor="black", linewidth=0.8, width=bar_width)

    # 3. Add connecting dashed lines between the bars
    for i in range(1, len(METEOR_SCORES)):
        # Draws a line from the top-right of the previous bar to the bottom-left of the current bar
        ax.plot(
            [x_positions[i-1] + (bar_width/2), x_positions[i] - (bar_width/2)], 
            [METEOR_SCORES[i-1], METEOR_SCORES[i-1]], 
            color="black", linestyle="--", linewidth=1, alpha=0.5
        )

    # 4. Automatically calculate and add text for values and +X% improvements
    for i, bar in enumerate(bars):
        current_val = METEOR_SCORES[i]
        text_str = f"{current_val:.2f}"
        
        # Calculate percentage growth from the PREVIOUS step
        if i > 0:
            prev_val = METEOR_SCORES[i-1]
            pct_change = ((current_val - prev_val) / prev_val) * 100
            if pct_change > 0:
                text_str += f"\n(+{pct_change:.1f}%)"
            elif pct_change < 0:
                text_str += f"\n({pct_change:.1f}%)"

        # Place the text slightly above the top of the current total
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            current_val + (max(METEOR_SCORES) * 0.02), 
            text_str,
            ha="center",
            va="bottom",
            fontsize=11,
            fontweight="bold",
            color="black"
        )

    # 5. Apply typography and styling
    ax.set_ylabel("METEOR Score", fontsize=14, fontweight="bold")
    ax.set_title("Pipeline Improvements: METEOR Score Progression", fontsize=16, fontweight="bold", pad=20)
    
    ax.set_xticks(x_positions)
    ax.set_xticklabels(labels, fontsize=12, fontweight="bold")
    plt.yticks(fontsize=12)
    
    # Dashed grid behind the bars
    ax.grid(axis="y", linestyle="--", alpha=0.3)
    ax.set_axisbelow(True)
    
    # Give the y-axis some headroom so the text doesn't get cut off
    ax.set_ylim(0, max(METEOR_SCORES) * 1.20)
    
    plt.tight_layout()
    plt.savefig("waterfall_improvement_meteor.png", dpi=300, bbox_inches="tight")
    plt.show()

if __name__ == "__main__":
    generate_waterfall_improvement_plot()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def generate_radar_chart():
    # =========================================================================
    # 1. ⬇️ PUT YOUR ACTUAL SCORES HERE ⬇️
    # Order: BLEU-1, BLEU-2, METEOR, ROUGE-L
    # =========================================================================
    metrics = ["BLEU-1", "BLEU-2", "METEOR", "ROUGE-L", "TIME-EPOCH"]  # Add more metrics if needed
    
    # Replace these with your actual numbers
    BASELINE_SCORES = [10.15, 3.66, 6.79, 18.51, 20.00]  
    BEST_SCORES = [58.01, 35.74, 35.90, 40.85, 13.10]    

    # Established pastel aesthetic
    color_baseline = "#FF7F6E" # Soft Coral
    color_best = "#8ED6A8"     # Soft Green

    # 2. Mathematical setup for the radar chart
    # We need to calculate the angle for each metric on the circle
    angles = np.linspace(0, 2 * np.pi, len(metrics), endpoint=False).tolist()

    # To close the plot, we need to append the first value to the end of the lists
    BASELINE_SCORES += BASELINE_SCORES[:1]
    BEST_SCORES += BEST_SCORES[:1]
    angles += angles[:1]
    
    # 3. Create the plot with a polar projection
    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
    
    # Rotate so the first metric is at the very top
    ax.set_theta_offset(np.pi / 2)
    ax.set_theta_direction(-1)

    # 4. Plot Baseline
    ax.plot(angles, BASELINE_SCORES, color=color_baseline, linewidth=2, linestyle='solid', label="Baseline")
    ax.fill(angles, BASELINE_SCORES, color=color_baseline, alpha=0.4)

    # 5. Plot Best Model
    ax.plot(angles, BEST_SCORES, color=color_best, linewidth=2, linestyle='solid', label="Best Model")
    ax.fill(angles, BEST_SCORES, color=color_best, alpha=0.4)

    # 6. Apply typography and styling
    # Set the labels for each axis
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(metrics, fontsize=14, fontweight="bold")
    
    # Customize the radial grids (the circular lines)
    plt.yticks(fontsize=11, fontweight="bold", color="grey")
    ax.grid(color="grey", linestyle="--", alpha=0.3)
    
    # Optional: Automatically scale the Y-axis to the max value, rounded up nicely
    max_val = max(max(BASELINE_SCORES), max(BEST_SCORES))
    ax.set_ylim(0, max_val * 1.1)

    # Add Title and Legend
    plt.title("Baseline vs. Best Model Performance", fontsize=16, fontweight="bold", pad=30)
    ax.legend(loc='upper right', bbox_to_anchor=(1.2, 1.1), fontsize=12, framealpha=0.9, edgecolor="black")

    plt.tight_layout()
    plt.savefig("radar_chart_comparison.png", dpi=300, bbox_inches="tight")
    plt.show()

if __name__ == "__main__":
    generate_radar_chart()
```
